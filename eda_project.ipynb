```python
#Imputing missing values intop bedrooms with median value

df['total_bedrooms'] = df['total_bedrooms'].fillna(df['total_bedrooms'].median())
```

```python
#

Hist_plot = sns.histplot(data = df, x= 'median_house_value', binwidth=50000)

for p in Hist_plot.patches:
    Hist_plot.text(p.get_x() + p.get_width()/2, p.get_height(), str(int(p.get_height())), ha = 'center', va = 'bottom')

plt.title('Distribution of median_house_value ')
plt.xlabel('median Value')
plt.ylabel('')
plt.yticks([])
plt.gca().spines[['top', 'right']].set_visible(False)
```

```python
# how many housing complexes fall in each ocean proximity category

plt.figure(figsize=(8,5))
Count_plot = sns.countplot(data = df, x= 'ocean_proximity', order =df['ocean_proximity'].value_counts().index)

for p in Count_plot.patches:
    Count_plot.text(p.get_x() + p.get_width()/2, p.get_height(), str(int(p.get_height())), ha = 'center', va = 'bottom')

plt.title('Count of Housing complexes by Ocean Proximity')
plt.xlabel('Ocean_proximity')
plt.ylabel('Hosuing complexes')

plt.gca().spines[['top', 'right']].set_visible(False)

plt.show()
```

```python
#outlier detection using boxplot

fig, axs = plt.subplots(nrows = 3, ncols = 3, figsize = (12,9) )

sns.boxplot(data = df, x = 'housing_median_age', ax = axs[0][0])

axs[0][0].set_title('Housing Median Age')
axs[0][0].set_xlabel('')
axs[0][0].set_ylabel('')
axs[0][0].spines[['top', 'right']].set_visible(False)


sns.boxplot(data = df, x = 'total_rooms', ax = axs[0][1])

axs[0][1].set_title('Total Rooms')
axs[0][1].set_xlabel('')
axs[0][1].set_ylabel('')
axs[0][1].spines[['top', 'right']].set_visible(False)

sns.boxplot(data = df, x = 'total_bedrooms', ax = axs[0][2])

axs[0][2].set_title('Total Bedrooms')
axs[0][2].set_xlabel('')
axs[0][2].set_ylabel('')
axs[0][2].spines[['top', 'right']].set_visible(False)

sns.boxplot(data = df, x = 'households', ax = axs[1][0])

axs[1][0].set_title('Total Housholds')
axs[1][0].set_xlabel('')
axs[1][0].set_ylabel('')
axs[1][0].spines[['top', 'right']].set_visible(False)

sns.boxplot(data = df, x = 'population', ax = axs[1][1])

axs[1][1].set_title('Total Population')
axs[1][1].set_xlabel('')
axs[1][1].set_ylabel('')
axs[1][1].spines[['top', 'right']].set_visible(False)

sns.boxplot(data = df, x = 'median_income', ax = axs[1][2])

axs[1][2].set_title('Total median_income')
axs[1][2].set_xlabel('')
axs[1][2].set_ylabel('')
axs[1][2].spines[['top', 'right']].set_visible(False)

sns.boxplot(data = df, x = 'median_house_value', ax = axs[2][0])

axs[2][0].set_title('Total median_house_value')
axs[2][0].set_xlabel('')
axs[2][0].set_ylabel('')
axs[2][0].spines[['top', 'right']].set_visible(False)

axs[2][1].set_visible(False)
axs[2][2].set_visible(False)

plt.tight_layout()

plt.show()
```
